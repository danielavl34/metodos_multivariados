## Implementar Regresión Lineal Simple usando NumPy

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
```

##Dataset sintético: y = 2x + 3 + $\epsilon$, con n = 100 de observaciones


```python
n = 100
X = np.random.rand(n, 1) * 10
e = np.random.rand(n, 1) * 1.5

Y = 2*X+ 3 * e;
print(Y)
```

```python
# 1. Varianza de X y de Y
var_X = np.var(X)
var_Y = np.var(Y)
print(f"Varianza de X = {var_X}")
print(f"Varianza de Y = {var_Y}")
```

```python
# 2. Valores de los coeficientes para la MCO
X_mean = np.mean(X) #Media de X
Y_mean = np.mean(Y) #Media de Y

SXX = np.sum((X - X_mean)**2) #Suma de cuadrados de X
SXY = np.sum((X - X_mean) * (Y - Y_mean)) #Suma de productos

print(f"Suma de cuadrados de X (SXX)= {SXX}")
print(f"Suma de productos (SXY)= {SXY}")
```

```python
# 3. Valores de B0 (intersección) y B1 (pendiente)
B1 = SXY / SXX
B0 = (Y_mean) - ((B1)*(X_mean))
print(f"β0 (intersección) = {B0}")
print(f"β1 (pendiente)    = {B1}")
```

```python
# 4. Valores de R^2 y MSE

#Predicciones del modelo
Y_pred = B0 + (B1 * X)

SST = np.sum((Y - Y_mean)**2) #Suma de cuadrados

SSR = np.sum((Y_pred - Y_mean)**2) #Variabilidad total

SSE = np.sum((Y - Y_pred)**2) #Variabilidad explicada

#Variabilidad residual
R2 = SSR / SST
MSE = np.mean((Y - Y_pred)**2)

print(f"R²  = {R2}")
print(f"MSE = {MSE}")
```

```python
# 5. Visualización de los resultados
plt.figure(figsize=(10, 6))
plt.scatter(X, Y, alpha=0.6, color='green', label='Datos observados')
plt.plot(X, Y_pred, color='purple', linewidth=2,label=f'Recta ajustada: y = {B0:3f} + {B1:3f}x')
plt.xlabel('X (Variable predictora)')
plt.ylabel('Y (Variable respuesta)')
plt.legend()
plt.show()
```
